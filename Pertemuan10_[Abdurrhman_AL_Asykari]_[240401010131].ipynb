```python
# ==========================================
# LANGKAH 1: Muat dan Eksplorasi Data
# ==========================================
import pandas as pd

df = pd.read_csv("telco_churn.csv")
print("Dimensi Data:", df.shape)
print("\nDistribusi Target (Churn):")
print(df["Churn"].value_counts(normalize=True))


# ==========================================
# LANGKAH 2: Preprocessing
# ==========================================
from sklearn.model_selection import train_test_split

# Hapus kolom ID jika ada agar tidak ikut ter-encode
if "customerID" in df.columns:
    df = df.drop(columns=["customerID"])

# Encoding fitur kategorikal (mengubah teks menjadi numerik)
df_encoded = pd.get_dummies(df, drop_first=True)

# Pisahkan X (fitur) dan y (target = Churn) secara dinamis
target_col = [col for col in df_encoded.columns if col.startswith("Churn")][0]
X = df_encoded.drop(columns=[target_col])
y = df_encoded[target_col]

# Split data training dan testing (dengan stratifikasi)
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)


# ==========================================
# LANGKAH 3: Latih Model
# ==========================================
from sklearn.ensemble import RandomForestClassifier

# Inisialisasi dan latih model Random Forest
rf = RandomForestClassifier(
    n_estimators=300, class_weight="balanced", random_state=42
)
rf.fit(X_tr, y_tr)


# ==========================================
# LANGKAH 4: Evaluasi Model
# ==========================================
from sklearn.metrics import classification_report, roc_auc_score

# Hitung prediksi kelas dan probabilitas
y_pred = rf.predict(X_te)
y_prob = rf.predict_proba(X_te)[:, 1]  # Probabilitas kelas churn (kelas 1)

# Tampilkan classification report dan ROC-AUC
print("\n--- Classification Report ---")
print(classification_report(y_te, y_pred))

roc_auc = roc_auc_score(y_te, y_prob)
print(f"ROC-AUC Score: {roc_auc:.4f}")


# ==========================================
# LANGKAH 5: Prediksi Probabilitas & Kesimpulan
# ==========================================
# Hitung probabilitas churn untuk data testing
probabilities = rf.predict_proba(X_te)[:, 1]

# Kesimpulan
print("\n--- Kesimpulan Temuan Utama Model ---")
print(
    "1. Model Random Forest berhasil dilatih dengan parameter class_weight='balanced' untuk mengatasi ketidakseimbangan data churn."
)
print(
    "2. Berdasarkan skor ROC-AUC, model memiliki kemampuan yang baik dalam membedakan antara pelanggan yang akan churn dan yang tidak."
)
print(
    "3. Probabilitas churn individual dapat digunakan oleh tim bisnis untuk mengidentifikasi pelanggan berisiko tinggi secara proaktif."
)
print(
    "4. Metrik recall pada laporan klasifikasi menjadi fokus penting agar sebagian besar pelanggan yang benar-benar ingin berhenti dapat terdeteksi sejak dini."
)
```
